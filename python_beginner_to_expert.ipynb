## 1.1 Variables

Variables are **containers** that store data values. Python is **dynamically typed** — you don't need to declare the type.

````python
# Variable assignment - no type declaration needed
x = 10              # integer
name = "Samrat"      # string
price = 99.5         # float
is_active = True     # boolean

print("x =", x)
print("name =", name)
print("price =", price)
print("is_active =", is_active)
````

````python
# Multiple assignment
a, b, c = 1, 2, 3
print(a, b, c)

# Swap variables (Pythonic way)
a, b = b, a
print("After swap:", a, b)
````

````python
# Variable naming rules:
# 1. Must start with letter or underscore
# 2. Cannot start with a number
# 3. Can only contain alphanumeric characters and underscores
# 4. Case-sensitive (age, Age, AGE are different)

my_variable = 10     # snake_case (Python convention)
_private = "hidden"   # leading underscore = private by convention
MAX_SIZE = 100        # ALL_CAPS = constant by convention

print(my_variable, _private, MAX_SIZE)
````

## 1.3 Data Types

Python has several built-in data types:

| Type | Example | Description |
|------|---------|-------------|
| `int` | `10` | Integer numbers |
| `float` | `3.14` | Decimal numbers |
| `str` | `"hello"` | Text / string |
| `bool` | `True/False` | Boolean |

| `tuple` | `(1, 2, 3)` | Ordered, immutable collection |

## 1.4 Operators

````python
# Arithmetic Operators
a, b = 17, 5

print(f"a + b  = {a + b}")    # Addition: 22
print(f"a - b  = {a - b}")    # Subtraction: 12
print(f"a * b  = {a * b}")    # Multiplication: 85
print(f"a / b  = {a / b}")    # Division (float): 3.4
print(f"a // b = {a // b}")   # Floor division: 3
print(f"a % b  = {a % b}")    # Modulus (remainder): 2
print(f"a ** b = {a ** b}")   # Exponentiation: 1419857
````

# Topic 2: IF / ELSE – Conditional Statements

Used to **make decisions** in your program. Python uses **indentation** (not `{}` like Java/C++).

### Syntax
```python
if condition:
    # code block (indented)
elif another_condition:
    # code block
else:
    # code block
```

````python
# Basic if/elif/else
marks = 75

if marks >= 90:
    print("Grade A - Excellent!")
elif marks >= 80:
    print("Grade B - Very Good")
elif marks >= 60:
    print("Grade C - Good")
elif marks >= 40:
    print("Grade D - Pass")
else:
    print("Grade F - Fail")
````

````python
# Nested if statements
age = 25
has_license = True

if age >= 18:
    if has_license:
        print("You can drive!")
    else:
        print("Get a license first.")
else:
    print("Too young to drive.")
````

````python
# Logical operators in conditions
temp = 25

if temp >= 20 and temp <= 30:
    print("Pleasant weather")

# Same thing, Pythonic way:
if 20 <= temp <= 30:
    print("Pleasant weather (chained comparison)")
````

````python
# Ternary operator (one-liner if/else)
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)

# Practical: get absolute value
num = -15
absolute = num if num >= 0 else -num
print(f"Absolute of {num} is {absolute}")
````

````python
# Truthy and Falsy values
# Falsy: 0, 0.0, "", [], {}, (), set(), None, False
# Everything else is Truthy

values = [0, 1, "", "hello", [], [1], None, True, False]

for v in values:
    if v:
        print(f"{str(v):>10} -> Truthy")
    else:
        print(f"{str(v):>10} -> Falsy")
````

````python
# match-case (Python 3.10+ - like switch)
command = "start"

match command:
    case "start":
        print("Starting...")
    case "stop":
        print("Stopping...")
    case "pause":
        print("Pausing...")
    case _:           # default case
        print("Unknown command")
````

# Topic 3: Loops & Functions

## 3.1 Loops

Loops are used to **repeat work**. Python has two main loop types:
- `for` loop: iterate over a sequence
- `while` loop: repeat while condition is true

````python
# for loop with range()
# range(start, stop, step) - stop is exclusive

print("1 to 5:")
for i in range(1, 6):
    print(i, end=" ")
print()

print("\nEven numbers 0-10:")
for i in range(0, 11, 2):
    print(i, end=" ")
print()

print("\nCountdown:")
for i in range(5, 0, -1):
    print(i, end=" ")
print()
````

````python
# for loop over collections
fruits = ["apple", "banana", "cherry"]

# Basic iteration
for fruit in fruits:
    print(fruit)

print()

# With index using enumerate()
for i, fruit in enumerate(fruits):
    print(f"{i}: {fruit}")

print()

# Iterating over a string
for char in "Python":
    print(char, end="-")
print()

# Iterating over a dictionary
person = {"name": "Samrat", "age": 25, "city": "Delhi"}
for key, value in person.items():
    print(f"{key}: {value}")
````

````python
# while loop
i = 1
while i <= 5:
    print(i, end=" ")
    i += 1
print()

# While with break and continue
print("\nSkip 3, stop at 7:")
for i in range(1, 11):
    if i == 3:
        continue   # skip this iteration
    if i == 7:
        break      # exit the loop
    print(i, end=" ")
print()
````

````python
# for-else: else runs when loop completes WITHOUT break
numbers = [2, 4, 6, 8]

for n in numbers:
    if n % 2 != 0:
        print(f"Found odd number: {n}")
        break
else:
    print("All numbers are even!")
````

````python
# List Comprehension (Pythonic loop-in-one-line)
squares = [x**2 for x in range(1, 6)]
print("Squares:", squares)

evens = [x for x in range(1, 21) if x % 2 == 0]
print("Evens:", evens)

# Nested comprehension
pairs = [(x, y) for x in range(3) for y in range(3) if x != y]
print("Pairs:", pairs)
````

## 3.2 Functions

Functions let you **reuse logic**. Define once, call many times.

````python
# Basic function
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

result = add(10, 20)
print("Sum:", result)
````

````python
# Default parameters
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Samrat"))              # uses default
print(greet("Samrat", "Namaste"))   # custom greeting
````

````python
# *args (variable number of arguments)
def total(*numbers):
    return sum(numbers)

print(total(1, 2, 3))        # 6
print(total(10, 20, 30, 40)) # 100

# **kwargs (keyword arguments)
def info(**details):
    for key, value in details.items():
        print(f"{key}: {value}")

info(name="Samrat", age=25, city="Delhi")
````

````python
# Lambda (anonymous) functions
square = lambda x: x ** 2
print(square(5))  # 25

# Lambda with sort
students = [("Alice", 85), ("Bob", 72), ("Charlie", 91)]
students.sort(key=lambda s: s[1], reverse=True)
print("Sorted by marks:", students)
````

````python
# Multiple return values
def min_max(arr):
    return min(arr), max(arr)

lo, hi = min_max([5, 3, 8, 1, 9])
print(f"Min: {lo}, Max: {hi}")
````

````python
# Scope: local vs global
x = 100  # global

def modify():
    x = 50  # local — does NOT change global x
    print("Inside function:", x)

modify()
print("Outside function:", x)

def modify_global():
    global x
    x = 50  # now changes global

modify_global()
print("After modify_global:", x)
````

# Topic 4: Pattern Problems & Time Complexity

## 4.1 Pattern Problems

Pattern problems build your **loop thinking**. They use nested loops.

````python
# Pattern 1: Right triangle of stars
n = 5
print("Pattern 1: Right triangle")
for i in range(1, n + 1):
    print("*" * i)
````

````python
# Pattern 2: Inverted triangle
print("Pattern 2: Inverted triangle")
for i in range(n, 0, -1):
    print("*" * i)
````

````python
# Pattern 3: Number triangle
print("Pattern 3: Number triangle")
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
````

````python
# Pattern 4: Pyramid
print("Pattern 4: Pyramid")
for i in range(1, n + 1):
    spaces = " " * (n - i)
    stars = "*" * (2 * i - 1)
    print(spaces + stars)
````

````python
# Pattern 5: Diamond
print("Pattern 5: Diamond")
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
````

````python
# Pattern 6: Pascal's Triangle
print("Pattern 6: Pascal's Triangle")
for i in range(n):
    row = [1]
    for j in range(1, i + 1):
        row.append(row[-1] * (i - j + 1) // j)
    print(" " * (n - i), end="")
    print("  ".join(f"{x:>2}" for x in row))
````

## 4.2 Time Complexity (Big-O Notation)

Shows **how fast code grows** as input size increases.

| Complexity | Name | Example |
|-----------|------|---------|
| O(1) | Constant | Array access `arr[0]` |
| O(log n) | Logarithmic | Binary search |
| O(n) | Linear | Single loop |
| O(n log n) | Linearithmic | Merge sort, Tim sort |
| O(n²) | Quadratic | Nested loop |
| O(2ⁿ) | Exponential | Recursive subsets |
| O(n!) | Factorial | Permutations |

````python
import time

# O(1) - Constant time
def constant_example(arr):
    return arr[0]  # always one operation

# O(n) - Linear time
def linear_example(n):
    total = 0
    for i in range(n):    # n operations
        total += i
    return total

# O(n²) - Quadratic time
def quadratic_example(n):
    count = 0
    for i in range(n):        # n
        for j in range(n):    # × n = n²
            count += 1
    return count

# O(log n) - Logarithmic time
def log_example(n):
    count = 0
    while n > 1:
        n //= 2    # halving each time
        count += 1
    return count

# Demonstrate growth
for n in [10, 100, 1000, 10000]:
    print(f"n={n:>5} | O(n)={n:>10} | O(n²)={n*n:>12} | O(log n)={log_example(n):>5}")
````

````python
# Timing comparison
import time

n = 5000

start = time.time()
linear_example(n)
linear_time = time.time() - start

start = time.time()
quadratic_example(n)
quad_time = time.time() - start

print(f"Linear   O(n)  for n={n}: {linear_time:.6f}s")
print(f"Quadratic O(n²) for n={n}: {quad_time:.6f}s")
print(f"Quadratic is ~{quad_time/max(linear_time, 0.000001):.0f}x slower")
````

## 5.2 2D Array (Matrix)

````python
# Creating 2D arrays
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Access: matrix[row][col]
print("Element [1][2]:", matrix[1][2])  # 6
print("First row:", matrix[0])           # [1, 2, 3]

# Get dimensions
rows = len(matrix)
cols = len(matrix[0])
print(f"Dimensions: {rows}x{cols}")
````

````python
# Creating 2D arrays dynamically
# CORRECT way:
rows, cols = 3, 4
grid = [[0] * cols for _ in range(rows)]
print("Correct grid:", grid)

# WRONG way (all rows share same reference!):
# bad_grid = [[0] * cols] * rows  # DON'T DO THIS

# Transpose a matrix
transposed = [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]
print("\nOriginal:")
for row in matrix:
    print(row)
print("Transposed:")
for row in transposed:
    print(row)
````

# Topic 7: Number System & Bit Manipulation

## 7.1 Number Systems

````python
# Number system conversions
num = 42

print(f"Decimal:     {num}")
print(f"Binary:      {bin(num)}")     # 0b101010
print(f"Octal:       {oct(num)}")     # 0o52
print(f"Hexadecimal: {hex(num)}")     # 0x2a

# Clean binary representation
print(f"Binary (clean): {num:08b}")   # 00101010

# Convert back to decimal
print(f"From binary:  {int('101010', 2)}")   # 42
print(f"From octal:   {int('52', 8)}")        # 42
print(f"From hex:     {int('2a', 16)}")        # 42
````

## 7.2 Bitwise Operators

| Operator | Symbol | Description |
|----------|--------|-------------|
| AND | `&` | Both bits must be 1 |
| OR | `\|` | At least one bit is 1 |
| XOR | `^` | Exactly one bit is 1 |
| NOT | `~` | Flip all bits |
| Left Shift | `<<` | Multiply by 2ⁿ |
| Right Shift | `>>` | Divide by 2ⁿ |

````python
a = 5   # 101 in binary
b = 3   # 011 in binary

print(f"a     = {a:04b} ({a})")
print(f"b     = {b:04b} ({b})")
print(f"a & b = {a & b:04b} ({a & b})")   # AND:  001 = 1
print(f"a | b = {a | b:04b} ({a | b})")   # OR:   111 = 7
print(f"a ^ b = {a ^ b:04b} ({a ^ b})")   # XOR:  110 = 6
print(f"~a    = {~a} (inverts all bits)")  # NOT
print(f"a << 1 = {a << 1:04b} ({a << 1})") # Left shift: 1010 = 10
print(f"a >> 1 = {a >> 1:04b} ({a >> 1})") # Right shift: 0010 = 2
````

````python
# Useful bit tricks

# Check if even/odd
n = 7
print(f"{n} is {'odd' if n & 1 else 'even'}")

# Check if power of 2
def is_power_of_two(n):
    return n > 0 and (n & (n - 1)) == 0

for i in [1, 2, 3, 4, 8, 16, 15]:
    print(f"{i:>3} is power of 2: {is_power_of_two(i)}")

# Count set bits (number of 1s)
def count_bits(n):
    count = 0
    while n:
        count += n & 1
        n >>= 1
    return count

print(f"\nBits in 13 ({13:b}): {count_bits(13)}")
print(f"Bits in 7 ({7:b}): {count_bits(7)}")
# Python built-in: bin(n).count('1')
````

# Topic 8: Interview Level – Strings & Bit Manipulation

Problem-solving applications for interviews.

````python
# Problem 1: Check Palindrome
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

# Two-pointer approach (interview preferred)
def is_palindrome_two_pointer(s):
    s = s.lower().replace(" ", "")
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True

tests = ["madam", "racecar", "hello", "A man a plan a canal Panama"]
for t in tests:
    print(f"'{t}' -> Palindrome: {is_palindrome(t)}")
````

````python
# Problem 2: Count character frequency
def char_frequency(s):
    freq = {}
    for ch in s.lower():
        if ch.isalpha():
            freq[ch] = freq.get(ch, 0) + 1
    return freq

text = "Hello World"
freq = char_frequency(text)
print(f"Frequency of '{text}':")
for ch, count in sorted(freq.items()):
    print(f"  '{ch}': {count}")
````

````python
# Problem 3: Check Anagram
def is_anagram(s1, s2):
    return sorted(s1.lower()) == sorted(s2.lower())

# Optimized with frequency count
from collections import Counter

def is_anagram_optimized(s1, s2):
    return Counter(s1.lower()) == Counter(s2.lower())

print(is_anagram("listen", "silent"))  # True
print(is_anagram("hello", "world"))    # False
````

````python
# Problem 4: First non-repeating character
def first_unique_char(s):
    freq = Counter(s)
    for i, ch in enumerate(s):
        if freq[ch] == 1:
            return i, ch
    return -1, None

idx, ch = first_unique_char("aabbcdd")
print(f"First unique: '{ch}' at index {idx}")  # 'c' at 4
````

````python
# Problem 5: Find unique element using XOR
# XOR properties: a ^ a = 0, a ^ 0 = a
def find_unique(arr):
    result = 0
    for num in arr:
        result ^= num
    return result

arr = [2, 3, 5, 3, 2]
print(f"Array: {arr}")
print(f"Unique element: {find_unique(arr)}")  # 5
````

````python
# Problem 6: Toggle case of each character
def toggle_case(s):
    return s.swapcase()  # built-in

# Manual approach using XOR with 32 (bit trick)
def toggle_case_bit(s):
    result = []
    for ch in s:
        if ch.isalpha():
            result.append(chr(ord(ch) ^ 32))  # XOR with 32 toggles case
        else:
            result.append(ch)
    return "".join(result)

print(toggle_case("Hello World"))      # hELLO wORLD
print(toggle_case_bit("Hello World"))  # hELLO wORLD
````

## 9.1 Linear Search — O(n)

````python
# Linear Search: check every element
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

arr = [5, 8, 2, 1, 9, 3]
target = 2
idx = linear_search(arr, target)
print(f"Found {target} at index {idx}")

idx = linear_search(arr, 99)
print(f"Found 99 at index {idx}")  # -1 (not found)
````

## 9.2 Binary Search — O(log n)

**Requires sorted array.** Halves the search space each step.

````python
# Binary Search: iterative
def binary_search(arr, target):
    low, high = 0, len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        
        if arr[mid] == target:
            return mid          # found!
        elif arr[mid] < target:
            low = mid + 1       # search right half
        else:
            high = mid - 1      # search left half

    return -1                   # not found

arr = [1, 3, 5, 7, 9, 11, 13, 15]
print(f"Search 7: index {binary_search(arr, 7)}")
print(f"Search 6: index {binary_search(arr, 6)}")
````

````python
# Binary Search: recursive
def binary_search_recursive(arr, target, low, high):
    if low > high:
        return -1
    
    mid = (low + high) // 2
    
    if arr[mid] == target:
        return mid
    elif arr[mid] < target:
        return binary_search_recursive(arr, target, mid + 1, high)
    else:
        return binary_search_recursive(arr, target, low, mid - 1)

print(f"Recursive search 11: {binary_search_recursive(arr, 11, 0, len(arr)-1)}")
````

# Topic 10: Sorting & Hashing

## 10.1 Sorting Algorithms

````python
# Bubble Sort — O(n²)
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break  # already sorted
    return arr

print("Bubble Sort:", bubble_sort([64, 34, 25, 12, 22, 11, 90]))
````

````python
# Selection Sort — O(n²)
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

print("Selection Sort:", selection_sort([64, 25, 12, 22, 11]))
````

````python
# Insertion Sort — O(n²) best case O(n)
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

print("Insertion Sort:", insertion_sort([12, 11, 13, 5, 6]))
````

````python
# Merge Sort — O(n log n)
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    result.extend(left[i:])
    result.extend(right[j:])
    return result

print("Merge Sort:", merge_sort([38, 27, 43, 3, 9, 82, 10]))
````

````python
# Quick Sort — O(n log n) average, O(n²) worst
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    return quick_sort(left) + middle + quick_sort(right)

print("Quick Sort:", quick_sort([3, 6, 8, 10, 1, 2, 1]))
````

````python
# Two Sum problem (classic interview question)
def two_sum(nums, target):
    seen = {}  # value -> index
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []

nums = [2, 7, 11, 15]
target = 9
print(f"Two Sum({nums}, {target}): {two_sum(nums, target)}")
````

## 11.1 Recursion

A function **calling itself**. Every recursion needs:
1. **Base case** (when to stop)
2. **Recursive case** (the function calls itself with a smaller problem)

````python
# Factorial: n! = n × (n-1)!
def factorial(n):
    if n == 0:           # base case
        return 1
    return n * factorial(n - 1)  # recursive case

print(f"5! = {factorial(5)}")   # 120

# Trace the calls:
# factorial(5) = 5 * factorial(4)
#              = 5 * 4 * factorial(3)
#              = 5 * 4 * 3 * factorial(2)
#              = 5 * 4 * 3 * 2 * factorial(1)
#              = 5 * 4 * 3 * 2 * 1 * factorial(0)
#              = 5 * 4 * 3 * 2 * 1 * 1 = 120
````

````python
# Sum of digits
def sum_digits(n):
    if n < 10:
        return n
    return (n % 10) + sum_digits(n // 10)

print(f"Sum of digits of 1234: {sum_digits(1234)}")  # 10

# Power function
def power(base, exp):
    if exp == 0:
        return 1
    return base * power(base, exp - 1)

print(f"2^10 = {power(2, 10)}")  # 1024
````

````python
# Stack using list
stack = []

# Push
stack.append(10)
stack.append(20)
stack.append(30)
print("Stack:", stack)

# Pop (remove last)
top = stack.pop()
print(f"Popped: {top}, Stack: {stack}")

# Peek (look at top without removing)
print(f"Top: {stack[-1]}")
print(f"Empty? {len(stack) == 0}")

# Application: Check balanced parentheses
def is_balanced(s):
    stack = []
    pairs = {')': '(', '}': '{', ']': '['}
    for ch in s:
        if ch in '({[':
            stack.append(ch)
        elif ch in ')}]':
            if not stack or stack[-1] != pairs[ch]:
                return False
            stack.pop()
    return len(stack) == 0

print("\n'({[]})' balanced:", is_balanced("({[]})"))  # True
print("'({[})' balanced:", is_balanced("({[})"))      # False
````

````python
# Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
    
    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node
    
    def display(self):
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next
        print(" -> ".join(elements) + " -> None")
    
    def reverse(self):
        prev = None
        current = self.head
        while current:
            next_node = current.next
            current.next = prev
            prev = current
            current = next_node
        self.head = prev

ll = LinkedList()
for val in [10, 20, 30, 40]:
    ll.append(val)

print("Original:")
ll.display()

ll.reverse()
print("Reversed:")
ll.display()
````

````python
# Binary Tree
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

# Build a tree:    1
#                 / \
#                2   3
#               / \
#              4   5
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

# Traversals
def inorder(node):
    if node:
        inorder(node.left)
        print(node.val, end=" ")
        inorder(node.right)

def preorder(node):
    if node:
        print(node.val, end=" ")
        preorder(node.left)
        preorder(node.right)

def postorder(node):
    if node:
        postorder(node.left)
        postorder(node.right)
        print(node.val, end=" ")

print("Inorder:  ", end=""); inorder(root); print()
print("Preorder: ", end=""); preorder(root); print()
print("Postorder:", end=""); postorder(root); print()
````

# Topic 12: Interview Level – Searching

Advanced binary search problems.

````python
# Problem 1: First and Last Occurrence
def first_occurrence(arr, target):
    low, high = 0, len(arr) - 1
    result = -1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            result = mid
            high = mid - 1  # keep searching left
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return result

def last_occurrence(arr, target):
    low, high = 0, len(arr) - 1
    result = -1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            result = mid
            low = mid + 1   # keep searching right
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return result

arr = [1, 2, 2, 2, 3, 4, 5]
print(f"First occurrence of 2: {first_occurrence(arr, 2)}")  # 1
print(f"Last occurrence of 2: {last_occurrence(arr, 2)}")    # 3
print(f"Count of 2: {last_occurrence(arr, 2) - first_occurrence(arr, 2) + 1}")  # 3
````

````python
# Problem 2: Find Peak Element
# A peak is greater than its neighbors
def find_peak(arr):
    low, high = 0, len(arr) - 1
    while low < high:
        mid = (low + high) // 2
        if arr[mid] < arr[mid + 1]:
            low = mid + 1   # peak is to the right
        else:
            high = mid       # peak is at mid or left
    return low

arr = [1, 3, 20, 4, 1, 0]
peak_idx = find_peak(arr)
print(f"Array: {arr}")
print(f"Peak at index {peak_idx}, value {arr[peak_idx]}")
````

````python
# Problem 3: Search in Rotated Sorted Array
# [4, 5, 6, 7, 0, 1, 2] — sorted but rotated
def search_rotated(arr, target):
    low, high = 0, len(arr) - 1
    
    while low <= high:
        mid = (low + high) // 2
        
        if arr[mid] == target:
            return mid
        
        # Left half is sorted
        if arr[low] <= arr[mid]:
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        # Right half is sorted
        else:
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
    
    return -1

arr = [4, 5, 6, 7, 0, 1, 2]
print(f"Array: {arr}")
print(f"Search 0: index {search_rotated(arr, 0)}")  # 4
print(f"Search 6: index {search_rotated(arr, 6)}")  # 2
print(f"Search 3: index {search_rotated(arr, 3)}")  # -1
````

````python
# Problem 4: Square Root using Binary Search
def sqrt_binary(n):
    if n < 2:
        return n
    low, high = 1, n // 2
    while low <= high:
        mid = (low + high) // 2
        if mid * mid == n:
            return mid
        elif mid * mid < n:
            low = mid + 1
            result = mid
        else:
            high = mid - 1
    return result

for n in [4, 8, 16, 25, 26]:
    print(f"sqrt({n}) = {sqrt_binary(n)}")
````

# Topic 13: Interview Level – Recursion

Classic recursion and backtracking problems.

````python
# Problem 1: Fibonacci (with memoization)

# Naive — O(2^n) — SLOW
def fib_naive(n):
    if n <= 1:
        return n
    return fib_naive(n - 1) + fib_naive(n - 2)

# Memoized — O(n) — FAST
def fib_memo(n, memo={}):
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fib_memo(n - 1, memo) + fib_memo(n - 2, memo)
    return memo[n]

# Using functools (Pythonic)
from functools import lru_cache

@lru_cache(maxsize=None)
def fib_cached(n):
    if n <= 1:
        return n
    return fib_cached(n - 1) + fib_cached(n - 2)

print("Fibonacci(10):", fib_memo(10))     # 55
print("Fibonacci(30):", fib_cached(30))   # 832040

# Compare speed
import time
start = time.time()
fib_naive(30)
print(f"\nNaive fib(30):  {time.time()-start:.4f}s")

start = time.time()
fib_memo(30)
print(f"Memo fib(30):   {time.time()-start:.6f}s")
````

````python
# Problem 2: Generate All Subsets (Power Set)
def subsets(nums):
    result = []
    
    def backtrack(start, current):
        result.append(current[:])  # add a copy
        for i in range(start, len(nums)):
            current.append(nums[i])
            backtrack(i + 1, current)
            current.pop()          # backtrack
    
    backtrack(0, [])
    return result

print("Subsets of [1, 2, 3]:")
for s in subsets([1, 2, 3]):
    print(f"  {s}")
````

````python
# Problem 3: Generate All Permutations
def permutations(nums):
    result = []
    
    def backtrack(current, remaining):
        if not remaining:
            result.append(current[:])
            return
        for i in range(len(remaining)):
            current.append(remaining[i])
            backtrack(current, remaining[:i] + remaining[i+1:])
            current.pop()  # backtrack
    
    backtrack([], nums)
    return result

print("Permutations of [1, 2, 3]:")
for p in permutations([1, 2, 3]):
    print(f"  {p}")
````

````python
# Problem 4: Tower of Hanoi
def tower_of_hanoi(n, source, auxiliary, target):
    if n == 1:
        print(f"  Move disk 1 from {source} to {target}")
        return
    tower_of_hanoi(n - 1, source, target, auxiliary)
    print(f"  Move disk {n} from {source} to {target}")
    tower_of_hanoi(n - 1, auxiliary, source, target)

print("Tower of Hanoi (3 disks):")
tower_of_hanoi(3, 'A', 'B', 'C')
````

````python
# Problem 5: N-Queens (Backtracking)
def solve_n_queens(n):
    solutions = []
    board = [['.' for _ in range(n)] for _ in range(n)]
    
    def is_safe(row, col):
        # Check column
        for i in range(row):
            if board[i][col] == 'Q':
                return False
        # Check upper-left diagonal
        i, j = row - 1, col - 1
        while i >= 0 and j >= 0:
            if board[i][j] == 'Q':
                return False
            i -= 1
            j -= 1
        # Check upper-right diagonal
        i, j = row - 1, col + 1
        while i >= 0 and j < n:
            if board[i][j] == 'Q':
                return False
            i -= 1
            j += 1
        return True
    
    def backtrack(row):
        if row == n:
            solutions.append([''.join(r) for r in board])
            return
        for col in range(n):
            if is_safe(row, col):
                board[row][col] = 'Q'
                backtrack(row + 1)
                board[row][col] = '.'  # backtrack
    
    backtrack(0)
    return solutions

solutions = solve_n_queens(4)
print(f"4-Queens: {len(solutions)} solutions found")
for i, sol in enumerate(solutions):
    print(f"\nSolution {i+1}:")
    for row in sol:
        print(f"  {row}")
````

````python
# Problem 6: Maze Solver (Backtracking)
def solve_maze(maze):
    rows, cols = len(maze), len(maze[0])
    path = []
    
    def backtrack(r, c):
        if r == rows - 1 and c == cols - 1:
            path.append((r, c))
            return True
        if r < 0 or r >= rows or c < 0 or c >= cols:
            return False
        if maze[r][c] != 1:
            return False
        
        maze[r][c] = 2  # mark visited
        path.append((r, c))
        
        # Try right, down, left, up
        if (backtrack(r, c+1) or backtrack(r+1, c) or
            backtrack(r, c-1) or backtrack(r-1, c)):
            return True
        
        path.pop()      # backtrack
        return False
    
    backtrack(0, 0)
    return path

maze = [
    [1, 0, 0, 0],
    [1, 1, 0, 1],
    [0, 1, 0, 0],
    [1, 1, 1, 1]
]

path = solve_maze([row[:] for row in maze])  # pass copy
print("Maze (1=open, 0=wall):")
for row in maze:
    print(f"  {row}")
print(f"\nPath found: {path}")
````
